```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Manual Linear Regression

```python
weeks_data = [1,2,3,4,5]
sales_data = [2,4,5,4,5]

df = pd.DataFrame({"Weeks":weeks_data,
                    "Sales":sales_data})

df
```

```python
X = df["Weeks"].values # Feature
Y = df["Sales"].values # Target
```

```python
mean_X = np.mean(X)
mean_Y = np.mean(Y)

n = len(X) # Banyak data atau minggu 1 - 5 berarty n = 5

kovarians = 0 # Ini yang (X - X bar)(Y - Y bar)
varians = 0   # Ini yang (X -X bar)^2

for i in range(n):
    kovarians += (X[i] - mean_X) * (Y[i] - mean_Y) # Ini yang (X - X bar)(Y - Y bar)
    varians += (X[i] - mean_X) ** 2                # Ini yang (X -X bar)^2

m = kovarians / varians
c = mean_Y - (m * mean_X)

print(f"m : {m} ")
print(f"c : {c} ")
```

```python
max_X = np.max(X) + 1
min_X = np.min(X) - 1

x = np.linspace(min_X, max_X)
y = (m * x) + c

plt.plot(x,y,color='#8c2500',label='Regression Line',linestyle='--')
plt.scatter(X,Y,color='#043078',label='Data Point')
plt.xlabel('Weeks')
plt.ylabel('Sales')
plt.title('LR Demo')
plt.grid(True)
plt.legend(loc='best')
```

## R2 atau R - Square

```python
rss = 0 # Residual Sum of Squares (RSS)
tss = 0 # Total Sum of Squares (TSS)

for i in range(n):
    y_pred = (m * X[i]) + c
    rss += (Y[i] - y_pred) ** 2 # (y-y ped)^2
    tss += (Y[i] - mean_Y) ** 2 # (y - y bar)^2

r2 = 1 - (rss/tss)
print(f"r2 : {r2}")
```

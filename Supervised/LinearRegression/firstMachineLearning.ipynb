```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

# Ambil dataset

```python
data = pd.read_csv("../../Datasets/headbrain.csv")
data.head(5)
```

## Features

```python
X = data['Head Size(cm^3)'].values
```

## Target

```python
Y = data['Brain Weight(grams)'].values
```

## Mengubah Feature jadi 2 dimensi

```python
X = X.reshape(-1,1)
```

#### Membagi Feature X dan Target Y yang _train dengan jumlah 80%, sedangkan di lakukan test dengan X dan Y yang _test sejumlah 20% karena test_size=0.20

```python
X_train, X_test, Y_train, Y_test = train_test_split( X, Y, test_size=0.20)
```

## Menyiapkan model untuk dilatih sedangkan lr.fit adalah proses melatih model

```python
lr = LinearRegression()
lr.fit(X_train, Y_train)
```

## Memberikan soal berupa X_test untuk diprediksi berapa berat otaknya

```python
y_pred = lr.predict(X_test)
```

### Mengambil score r^2 dari perbandingan antara Y_test atau data actual dengan y_pred atau data hasil prediksi

```python
r2 = r2_score(Y_test, y_pred)
r2
```

```python
plt.scatter(X_test, Y_test, color='black', label='Actual')
plt.plot(X_test, y_pred, color='red', label='Predicted')
plt.legend()
```
